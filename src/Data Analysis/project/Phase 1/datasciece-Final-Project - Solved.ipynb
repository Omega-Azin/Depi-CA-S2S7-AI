

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with ........... for numeric and ......for categorical

```python
# trying to fill missing values in my datset using for loop 

for col in gbf.columns:
    if gbf[col].isnull().sum() > 0:
        if pd.api.types.is_numeric_dtype(gbf[col]):
            # fill numeric columns with the median
            gbf[col] = gbf[col].fillna(gbf[col].median())
        elif col != 'member_gender':
            # fill categorical columns (except member_gender, handled separately) with the mode
            gbf[col] = gbf[col].fillna(gbf[col].mode()[0])
```

```python
#check the Nulls 

gbf.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
##  still got a missing value in the member_gender, so let's explore this column more 

gbf.member_gender.value_counts(dropna=False)
```

```python
# fill  dataset with missing values.

gbf['member_gender'] = gbf['member_gender'].fillna('Other')
```

# <a id="1"> 1 ) . What is the distribution for User type?</a>

```python
## I will import seaborn library to support me in making visualizations 

import seaborn as sns

plt.figure(figsize = (8,5), dpi = 100)

order = gbf['user_type'].value_counts().index
sns.countplot(data=gbf, x='user_type', color=sb.color_palette()[0], order=order)

plt.xlabel('User Type')
plt.ylabel('Frequency')
plt.title("Distribution of Users by User Type")

plt.show()
```

# <a id="3"> 3 ) What is the distribution for users' ages? </a>

```python
# Make visualization to easy get insight

plt.figure(figsize=(10,5), dpi=100)

bins = np.arange(gbf['age'].min(), gbf['age'].quantile(0.995)+2, 2)
plt.hist(data=gbf, x='age', bins=bins, color=sb.color_palette()[2])

plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title("Distribution of Users' Age")

plt.show()
```

> ### <mark>Your analsys and exploration here </mark>

The age distribution is **right-skewed**: most riders are between roughly 25 and 40 years old, with a peak around the early 30s. Fewer trips are made by older riders, and the distribution tails off gradually after age 45.

# <a id="14"> 14 )  .  What is the  Correlation between `specific numeric variables` in the dataset?</a>


```python
# Make visualization to easy get insight

selected_cols = ['duration_min', 'duration_hour', 'age', 'member_birth_year']
corr_selected = gbf[selected_cols].corr()

plt.figure(figsize=(6,5), dpi=100)
sns.heatmap(corr_selected, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Between Specific Numeric Variables')
plt.show()
```

> ### <mark>Your analsys and exploration here </mark>

Example usage
- There is `no Correlation` for any variable with time ( `duration sec /minutes`)


- `duration_min` and `duration_hour` are (as expected) perfectly correlated since one is a linear rescaling of the other.


- `age` and `member_birth_year` are strongly (negatively) correlated, again because `age` was derived directly from `member_birth_year`.
