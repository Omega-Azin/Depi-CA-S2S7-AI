```python
def info_user(name : str| None = "Shaltout", age : int| None = 0) -> str:
    '''
    This function takes a name and age as input and returns a formatted string.
    Args:
        name (str): The name of the user.
        age (int): The age of the user.
    Returns:
        str: A formatted string with the user's name and age.
    '''
    return f"My name is {name} and I am {age} years old."
```

```python
info_user("Youssef", age = 23)
```

```python
def train_model(model_name, *datasets, **hyperparameters):
    print(f"Training model: {model_name}")

    print("\nDatasets")
    for dataset in datasets:
        print("-", {dataset})

    print("\nHyperparameters:")
    for key, value in hyperparameters.items():
        print(f"{key}: {value}")

```

```python
train_model("Fraud detection","train.csv","validation.csv","test.csv",learning_rate=0.001,epochs=100,batch_size=64,optimizer="Youssef",early_stopping=True)
```

```python
transformer = {
"lower": lambda txt: txt.lower(),
"upper": lambda txt: txt.upper(),
"strip": lambda txt: txt.strip(),
"title ": lambda txt: txt.title(),
"reverse": lambda txt: txt[::-1]
}

txt = "  Machine Learning with Python  "
print(transformer["lower"](txt))
print(transformer["upper"](txt))
print(transformer["strip"](txt))
print(transformer["title "](txt))
print(transformer["reverse"](txt))
```
