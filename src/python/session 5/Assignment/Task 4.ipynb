### Task 1: Multiplication Table

```python
class MultiplicationTable:
    """MultiplicationTable generates and prints the multiplication table for a given number.

    Attributes:
        number (int): The base number for the multiplication table.
        limit (int): The upper range multiplier (default is 10).
    """

    def __init__(self, number: int, limit: int = 10):
        """Initializes the MultiplicationTable with a number and an optional range limit."""
        self.number = number
        self.limit = limit

    def generate_table(self):
        """Prints the multiplication table from 1 up to the specified limit."""
        for i in range(1, self.limit + 1):
            print(f"{self.number} * {i} = {self.number * i}")
```

```python
# Object Creation and Verification

table_obj = MultiplicationTable(number=9)
table_obj.generate_table()
```

### Task 2: Twin Primes

```python
class TwinPrimesFinder:
    """TwinPrimesFinder identifies and displays twin primes up to a given limit.

    Attributes:
        limit (int): The upper limit to search for twin prime pairs.
    """

    def __init__(self, limit: int = 1000):
        """Initializes the TwinPrimesFinder with the target upper bound."""
        self.limit = limit

    def _is_prime(self, n: int) -> bool:
        """Determines if a given number is prime.

        Args:
            n (int): The number to test.

        Returns:
            bool: True if prime, False otherwise.
        """
        if n < 2:
            return False
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                return False
        return True

    def find_twin_primes(self) -> list[tuple[int, int]]:
        """Finds all twin prime pairs less than the specified limit.

        Returns:
            list[tuple[int, int]]: A list of tuples containing twin prime pairs.
        """
        twin_primes = []
        for i in range(3, self.limit - 2, 2):
            if self._is_prime(i) and self._is_prime(i + 2):
                twin_primes.append((i, i + 2))
        return twin_primes
```

```python
# Object Creation and Verification

twin_primes_obj = TwinPrimesFinder(limit=1000)
results = twin_primes_obj.find_twin_primes()
for p1, p2 in results[:10]:  # Displaying the first 10 for demonstration
    print(f"{p1} and {p2}")
```

### Task 3: Prime Factors

```python
class PrimeFactorsCalculator:
    """PrimeFactorsCalculator extracts the prime factors of a target number.

    Attributes:
        number (int): The integer to be factorized.
    """

    def __init__(self, number: int):
        """Initializes the PrimeFactorsCalculator with a target integer."""
        self.number = number

    def get_prime_factors(self) -> list[int]:
        """Calculates and returns all prime factors of the stored number.

        Returns:
            list[int]: List containing the prime factors.
        """
        factors = []
        temp = self.number
        d = 2
        while d * d <= temp:
            while temp % d == 0:
                factors.append(d)
                temp //= d
            d += 1
        if temp > 1:
            factors.append(temp)
        return factors
```

```python
# Object Creation and Verification

factors_obj = PrimeFactorsCalculator(number=56)
print(f"Prime factors of 56: {factors_obj.get_prime_factors()}")
```

### Task 4: Decimal to Binary Converter

```python
class DecimalToBinaryConverter:
    """DecimalToBinaryConverter converts non-negative decimal integers into binary strings.

    Attributes:
        number (int): The decimal number to convert.
    """

    def __init__(self, number: int):
        """Initializes the DecimalToBinaryConverter with a decimal number."""
        self.number = number

    def convert(self) -> str:
        """Converts the integer to its binary string representation.

        Returns:
            str: Binary representation of the integer.
        """
        if self.number == 0:
            return "0"

        temp = self.number
        binary_str = ""
        while temp > 0:
            binary_str = str(temp % 2) + binary_str
            temp //= 2
        return binary_str
```

```python
# Object Creation and Verification

converter_obj = DecimalToBinaryConverter(number=11)
print(f"Decimal 11 to Binary: {converter_obj.convert()}")
```

### Task 5: Perfect Numbers Finder

```python
class PerfectNumbersFinder:
    """PerfectNumbersFinder detects perfect numbers within a defined numerical range.

    Attributes:
        start (int): The starting integer of the range.
        end (int): The ending integer of the range.
    """

    def __init__(self, start: int, end: int):
        """Initializes the range boundaries for perfect number calculation."""
        self.start = start
        self.end = end

    def _is_perfect(self, n: int) -> bool:
        """Checks if a number is equal to the sum of its proper divisors.

        Args:
            n (int): The number to evaluate.

        Returns:
            bool: True if perfect, False otherwise.
        """
        if n <= 1:
            return False
        divisors_sum = 1
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                divisors_sum += i
                if i != n // i:
                    divisors_sum += n // i
        return divisors_sum == n

    def get_perfect_numbers(self) -> list[int]:
        """Searches the specified range and returns a list of all perfect numbers found.

        Returns:
            list[int]: List of perfect numbers in range.
        """
        return [
            num
            for num in range(self.start, self.end + 1)
            if self._is_perfect(num)
        ]
```

```python
# Object Creation and Verification

perfect_obj = PerfectNumbersFinder(start=0, end=100)
print(f"Perfect numbers in range [0, 100]: {perfect_obj.get_perfect_numbers()}")
```

### Task 6: Interactive Simple Calculator

```python
class SimpleCalculator:
    """SimpleCalculator provides arithmetic operations via interactive console prompts."""

    def __init__(self):
        """Initializes the calculator status."""
        self.is_running = True

    def display_menu(self):
        """Prints the menu options for arithmetic operations."""
        print("Welcome to the Simple Calculator!")
        print("Select an operation:")
        print("1. Addition (+)")
        print("2. Subtraction (-)")
        print("3. Multiplication (*)")
        print("4. Division (/)")

    def run(self):
        """Executes the calculator loop, capturing user inputs and printing results."""
        self.display_menu()
        while self.is_running:
            choice = input(
                "Enter your choice (1/2/3/4) or 'exit' to quit: "
            ).strip()

            if choice.lower() == "exit":
                print("Exiting the calculator. Goodbye!")
                self.is_running = False
                break

            if choice in ["1", "2", "3", "4"]:
                try:
                    num1 = float(input("Enter first number: "))
                    num2 = float(input("Enter second number: "))
                except ValueError:
                    print("Invalid input. Please enter numbers.")
                    continue

                if choice == "1":
                    print(f"{num1} + {num2} = {num1 + num2}")
                elif choice == "2":
                    print(f"{num1} - {num2} = {num1 - num2}")
                elif choice == "3":
                    print(f"{num1} * {num2} = {num1 * num2}")
                elif choice == "4":
                    if num2 == 0:
                        print("Error! Division by zero.")
                    else:
                        print(f"{num1} / {num2} = {num1 / num2}")
            else:
                print("Invalid Choice. Please try again.")
```

```python
# Object Creation and Verification

calc_obj = SimpleCalculator()
calc_obj.run()
```
